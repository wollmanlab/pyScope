```python
from file_handler import FileHandler
import os
file_handler = FileHandler()
for filename in os.listdir(file_handler.system_state_dir):
    filepath = os.path.join(file_handler.system_state_dir, filename)
    if os.path.exists(filepath):
        os.remove(filepath)
        file_handler.log(f"Removed file: {filename}")
```

```python
from Scope.positions import Positions
from Scope.cyanscope import CyanScope as Scope
scope = Scope(enable_core=True)
# Create positions object
positions_obj = Positions(
    fov_info=scope.fov_info,
    offsets=scope.offsets,
    axis_mapping=scope.axis_mapping,
    limits=scope.limits,
    save_dir=file_handler.system_state_dir
)
# Add ALL wells from plate config
selected_plate = 'Underwood6'
group1_wells = ['A','B','C']
group2_wells = ['D','E','F']
selected_wells = group1_wells + group2_wells
fluidics_well_assignments = {well:well for well in selected_wells}
group_assignments = {well:None for well in selected_wells}
group_assignments.update({well:'Group 1' for well in group1_wells})
group_assignments.update({well:'Group 2' for well in group2_wells})
print(group_assignments)
plate_config = {well:selected_plate for well in selected_wells}
for well_name, well_config in plate_config.items():
    positions_obj.add_well(well_name, well_config)
file_handler.save_positions(positions_obj.positions.copy())
```
