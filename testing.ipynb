```python
""" Overhead Test """

X = scope.X
Y = scope.Y
Z = scope.Z
channels = ['FarRed', 'DeepBlue', 'Green', 'Orange']
state = scope.state.copy()
i=0
start = time.time()
step_size = 1000
for x in range(-1,2):
    for y in range(-1,2):
        state['X'] = X + x*step_size
        state['Y'] = Y + y*step_size
        # state['XY'] = (state['X'], state['Y'])
        for channel in channels:
            i+=1
            print(i,x,y,channel,state['X'] ,state['Y'] )
            state['Channel'] = channel
            scope.update_state(state=state.copy())
            img = scope.snapImage()
            print('---',scope.X, scope.Y, scope.Z)
print(time.time() - start)
print((time.time() - start)/i)
```
